# 8. Relation entre les variables explicatives et la cible

Cette étape consiste à analyser la relation entre les variables explicatives et la variable cible `Time_taken(min)`.

L'objectif est d'identifier les variables qui présentent les relations les plus importantes avec le temps total de livraison et celles qui semblent avoir une relation faible avec celui-ci.

Pour les variables numériques, nous utiliserons principalement la corrélation et les nuages de points (`scatterplots`). Pour les variables catégorielles, nous analyserons les statistiques du temps de livraison par catégorie ainsi que leurs distributions à l'aide de `boxplots`.

> **Remarque :** une relation statistique entre une variable et la cible ne signifie pas nécessairement qu'il existe une relation causale. Cette analyse permet principalement d'identifier les variables potentiellement utiles pour la modélisation.


## 8.1 Variables numériques vs cible

Les variables numériques retenues pour cette analyse sont :

* `Delivery_person_Age`
* `Delivery_person_Ratings`
* `Vehicle_condition`
* `multiple_deliveries`

Pour chaque variable, nous calculons son coefficient de corrélation avec `Time_taken(min)` et nous représentons graphiquement sa relation avec la cible.

La corrélation permet de mesurer la force et la direction d'une relation linéaire :

* une corrélation proche de `+1` indique une relation positive forte ;
* une corrélation proche de `-1` indique une relation négative forte ;
* une corrélation proche de `0` indique une faible relation linéaire.

La corrélation ne permet cependant pas de détecter toutes les relations non linéaires.


````python
print("\n-----------------------------------------------------------\n")

numeric_features = continuous_columns + discrete_columns

numeric_correlations = (
    df[numeric_features + [target_column]]
    .corr(numeric_only=True)[target_column]
    .drop(target_column)
    .sort_values(key=abs, ascending=False)
)

print("Corrélations avec la cible :")
display(numeric_correlations.to_frame("Correlation"))
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

for column in numeric_features:
    correlation = df[[column, target_column]].corr().iloc[0, 1]

    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x=column,
        y=target_column,
        alpha=0.25
    )

    sns.regplot(
        data=df,
        x=column,
        y=target_column,
        scatter=False,
        ci=None
    )

    plt.title(
        f"{column} vs {target_column} "
        f"(r = {correlation:.3f})"
    )

    plt.xlabel(column)
    plt.ylabel("Delivery time (min)")
    plt.tight_layout()
    plt.show()
````

## 8.2 Variables catégorielles vs cible

Les variables catégorielles sont analysées en comparant la distribution du temps de livraison entre leurs différentes catégories.

Pour chaque variable, nous calculons :

* le nombre d'observations ;
* la moyenne du temps de livraison ;
* la médiane ;
* l'écart-type.

Les `boxplots` permettent ensuite de visualiser la distribution de `Time_taken(min)` pour chaque catégorie et de comparer leur dispersion ainsi que leurs valeurs extrêmes.


````python
categorical_statistics = {}

for column in categorical_columns:
    stats = (
        df.groupby(column, dropna=False)[target_column]
        .agg(
            count="count",
            mean="mean",
            median="median",
            std="std",
            min="min",
            max="max"
        )
        .sort_values("mean")
    )

    categorical_statistics[column] = stats

    print(f"\n{'=' * 70}")
    print(f"{column}")
    print(f"{'=' * 70}")

    display(stats)
````

````python
for column in categorical_columns:

    print(f"\n{column} - distribution:")
    display(df[column].value_counts(dropna=False))

    plt.figure(figsize=(10, 6))

    sns.boxplot(
        data=df,
        x=column,
        y=target_column
    )

    plt.title(f"{column} vs {target_column}")
    plt.xlabel(column)
    plt.ylabel("Delivery time (min)")
    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()
````

## 8.3 Matrice de corrélation

La matrice de corrélation permet d'analyser simultanément les relations linéaires entre les variables numériques.

Elle permet notamment :

* d'identifier les variables les plus corrélées avec la cible ;
* de détecter d'éventuelles corrélations fortes entre variables explicatives ;
* d'identifier les variables pouvant présenter un problème de redondance.

Les coordonnées GPS sont incluses dans cette analyse descriptive. Cependant, leur interprétation individuelle reste limitée, car la distance entre le restaurant et le lieu de livraison est une information plus pertinente. Une variable `Distance_km` sera donc créée ultérieurement lors de la phase de feature engineering.


````python
correlation_columns = (
    continuous_columns
    + gps_columns
    + discrete_columns
    + [target_column]
)

correlation_matrix = df[correlation_columns].corr(numeric_only=True)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Matrice de corrélation des variables numériques")
plt.tight_layout()
plt.show()
````

````python
target_correlations = (
    correlation_matrix[target_column]
    .drop(target_column)
    .sort_values(key=abs, ascending=False)
)

display(
    target_correlations.to_frame("Correlation_with_target")
)
````

## 8.4 Identification des variables les plus liées à la cible

Les variables numériques sont classées selon la valeur absolue de leur coefficient de corrélation avec `Time_taken(min)`.

Pour les variables catégorielles, la comparaison est basée sur les différences observées entre les moyennes du temps de livraison des différentes catégories.

Cette analyse permet d'identifier les variables qui semblent les plus informatives pour la suite du projet.

Il faut toutefois distinguer une forte association statistique d'un véritable effet causal. La sélection finale des variables sera confirmée lors de la modélisation.


````python
strongest_numeric = (
    numeric_correlations
    .abs()
    .sort_values(ascending=False)
)

print("Variables numériques les plus liées à la cible :")
display(
    strongest_numeric.to_frame("Absolute_correlation")
)
````

````python
strongest_numeric_table = (
    numeric_correlations
    .to_frame("Correlation")
)

strongest_numeric_table["Absolute_correlation"] = (
    strongest_numeric_table["Correlation"].abs()
)

strongest_numeric_table = strongest_numeric_table.sort_values(
    "Absolute_correlation",
    ascending=False
)

display(strongest_numeric_table)
````

````python
categorical_relationships = []

for column in categorical_columns:

    stats = (
        df.groupby(column, dropna=False)[target_column]
        .mean()
        .dropna()
    )

    categorical_relationships.append({
        "Variable": column,
        "Minimum_mean": stats.min(),
        "Maximum_mean": stats.max(),
        "Mean_difference": stats.max() - stats.min()
    })

categorical_relationships = pd.DataFrame(
    categorical_relationships
).sort_values(
    "Mean_difference",
    ascending=False
)

display(categorical_relationships)
````

## 8.5 Identification des variables les moins liées à la cible

Les variables présentant les plus faibles associations avec `Time_taken(min)` sont également identifiées.

Pour les variables numériques, nous utilisons la valeur absolue de la corrélation.

Pour les variables catégorielles, nous utilisons la différence entre la moyenne maximale et la moyenne minimale observée entre les catégories.

Une faible association dans cette analyse ne signifie cependant pas nécessairement que la variable est inutile. Une variable peut avoir une relation non linéaire ou interagir avec d'autres variables et devenir utile pour un modèle de Machine Learning.


````python
weakest_numeric = (
    numeric_correlations
    .abs()
    .sort_values(ascending=True)
)

print("Variables numériques les moins liées à la cible :")

display(
    weakest_numeric.to_frame("Absolute_correlation")
)
````

````python
weakest_categorical = categorical_relationships.sort_values(
    "Mean_difference",
    ascending=True
)

print("Variables catégorielles présentant les plus faibles différences de moyenne :")

display(weakest_categorical)
````

### 8.6 — Visualisation globale des relations

## 12. Traitement et standardisation des chaînes de caractères

Les colonnes textuelles peuvent contenir des espaces inutiles, des préfixes parasites ou des valeurs représentant des données manquantes sous forme de chaînes de caractères.

Le nettoyage sera effectué en plusieurs étapes :

1. supprimer les espaces inutiles au début et à la fin des chaînes ;
2. supprimer le préfixe `conditions` de `Weatherconditions` ;
3. convertir les chaînes `"NaN"` en véritables valeurs manquantes (`NaN`) ;
4. vérifier les modalités restantes afin de détecter d'éventuelles catégories inattendues.

````python
print("\n-----------------------------------------------------------\n")
print("String cleaning")

for column in categorical_columns:
    df[column] = df[column].str.strip()

df["Weatherconditions"] = (
    df["Weatherconditions"]
    .str.replace("conditions", "", regex=False)
    .str.strip()
)

for column in categorical_columns:
    df[column] = df[column].replace("NaN", np.nan)

print("String cleaning completed.")

print("\nUnique values after cleaning:")

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
````

### 14.9 Validation des coordonnées GPS

Les coordonnées géographiques doivent respecter des contraintes physiques :

- latitude comprise entre -90° et 90° ;
- longitude comprise entre -180° et 180°.

Cependant, respecter ces limites ne garantit pas que la coordonnée soit réellement valide. Une paire `(0, 0)` peut par exemple correspondre à une valeur par défaut utilisée lorsqu'une position est inconnue.

Les coordonnées seront donc contrôlées afin d'identifier :
- les valeurs hors limites ;
- les coordonnées nulles `(0, 0)` ;
- les éventuelles coordonnées manifestement incohérentes entre le restaurant et le lieu de livraison.

Les valeurs invalides seront transformées en valeurs manquantes ou les observations concernées seront traitées selon leur impact sur les variables géographiques dérivées.

````python
print("\n-----------------------------------------------------------\n")
print("GPS validation")

latitude_columns = [
    "Restaurant_latitude",
    "Delivery_location_latitude",
]

longitude_columns = [
    "Restaurant_longitude",
    "Delivery_location_longitude",
]

for column in latitude_columns:
    invalid_count = (
        (df[column] < -90) |
        (df[column] > 90)
    ).sum()

    print(
        f"{column}: "
        f"{invalid_count} out-of-range values"
    )

for column in longitude_columns:
    invalid_count = (
        (df[column] < -180) |
        (df[column] > 180)
    ).sum()

    print(
        f"{column}: "
        f"{invalid_count} out-of-range values"
    )
    
zero_coordinates = (
    (df["Restaurant_latitude"] == 0) &
    (df["Restaurant_longitude"] == 0)
)

print(
    "Restaurant coordinates equal to (0, 0):",
    zero_coordinates.sum()
)

zero_delivery_coordinates = (
    (df["Delivery_location_latitude"] == 0) &
    (df["Delivery_location_longitude"] == 0)
)

print(
    "Delivery coordinates equal to (0, 0):",
    zero_delivery_coordinates.sum()
)

print("\n-----------------------------------------------------------\n")

print("Count of delivery persons with zero restaurant coordinates by vehicle type:")

df[(df["Restaurant_latitude"] == 0) &
    (df["Restaurant_longitude"] == 0)].groupby("Type_of_vehicle").agg({
        "Delivery_person_ID" : "count"
    }).sort_values("Delivery_person_ID", ascending=False)
````

## 16.2 Méthode de l’IQR

La méthode de l’IQR identifie comme aberrantes les valeurs situées en dehors des bornes suivantes :

- Borne inférieure : `Q1 - 1,5 × IQR`
- Borne supérieure : `Q3 + 1,5 × IQR`

````python
outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    
    IQR = q3 - q1
    
    lower_bound = q1 - 1.5 * IQR
    upper_bound = q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    outlier_summary.append({
        "column": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Number of Outliers": len(outliers),
        "Outliers Percentage": (len(outliers) / len(df)) * 100
    })
    
outlier_summary_df = pd.DataFrame(outlier_summary)

print("\n-----------------------------------------------------------\n")
print("Summary of Outliers:")
outlier_summary_df
````

## 16.3 Identification des valeurs aberrantes


````python
print("\n-----------------------------------------------------------\n")

columns_with_outliers = outlier_summary_df[outlier_summary_df["Number of Outliers"] > 0]

for col in columns_with_outliers["column"]:
    lower_bound = columns_with_outliers.loc[columns_with_outliers["column"] == col, "Lower Bound"].values[0]
    upper_bound = columns_with_outliers.loc[columns_with_outliers["column"] == col, "Upper Bound"].values[0]
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"\nOutliers in column '{col}':")
    display(outliers)

    print("\n-----------------------------------------------------------\n")
````

## 20. Conservation de `Time_Order_picked` et calcul du temps de préparation
 
`Time_Order_picked` était initialement supprimée pour éviter une fuite d'information, en supposant que le modèle devait prédire le temps de livraison **au moment où la commande est passée**.
 
Nouvelle décision : cette variable est conservée afin de calculer un **temps de préparation** (`Preparation_Time_min`), soit le délai entre la commande et sa prise en charge par le livreur. Cette information peut être très informative (temps cuisine, charge du restaurant, etc.).
 
⚠️ **Conséquence sur le cadrage du problème** : en utilisant `Time_Order_picked`, le modèle prédit désormais le temps de livraison **à partir du moment où la commande est récupérée**, et non plus au moment où elle est passée. Ce n'est pas un défaut, mais un changement de périmètre métier qu'il faut documenter (le use case devient : "combien de temps encore, une fois le livreur en route ?").
 
Le calcul du temps de préparation doit gérer le passage de minuit (ex. commande à 23h50, prise en charge à 00h05) :
 
```python
def time_diff_minutes(start, end):
    diff = (end - start).dt.total_seconds() / 60
    diff = np.where(diff < 0, diff + 24 * 60, diff)
    return diff
 
df["Preparation_Time_min"] = time_diff_minutes(
    df["Time_Orderd"], df["Time_Order_picked"]
)
 
print("\n-----------------------------------------------------------\n")
print("Statistiques du temps de préparation (minutes) :")
df["Preparation_Time_min"].describe()
```
 
`Time_Order_picked` est conservée dans le dataset (pour l'extraction ultérieure de `Pickup_Hour` par exemple), et `Preparation_Time_min` devient une nouvelle variable candidate pour la modélisation.

## 21. Traitement de `Time_Orderd`
 
`Time_Orderd` contient 1 731 valeurs manquantes (~3,80 %). Comme `Time_Order_picked` ne contient, elle, aucune valeur manquante, on peut désormais **imputer `Time_Orderd` à partir de `Time_Order_picked`** en utilisant le temps de préparation médian calculé sur les observations complètes — une imputation informée plutôt qu'une heure moyenne arbitraire.
 
Étapes :
1. Calculer `Preparation_Time_min` sur les lignes où `Time_Orderd` est renseignée (fait en section 20).
2. Calculer le temps de préparation médian sur ces lignes.
3. Pour les lignes où `Time_Orderd` est manquante, l'estimer comme `Time_Order_picked - médiane(Preparation_Time_min)`.
4. Conserver un indicateur `Time_Orderd_Was_Missing` pour tracer les valeurs imputées.
```python
print("\n-----------------------------------------------------------\n")
 
median_prep_time = df["Preparation_Time_min"].median()
print(f"Temps de préparation médian observé : {median_prep_time:.1f} minutes")
 
missing_order_time = df["Time_Orderd"].isna()
df["Time_Orderd_Was_Missing"] = missing_order_time.astype(int)
 
print(f"Nombre de valeurs de Time_Orderd à imputer : {missing_order_time.sum()}")
 
df.loc[missing_order_time, "Time_Orderd"] = (
    df.loc[missing_order_time, "Time_Order_picked"]
    - pd.to_timedelta(median_prep_time, unit="m")
)

df["Preparation_Time_min"] = time_diff_minutes(
    df["Time_Orderd"], df["Time_Order_picked"]
)
 
print("\nValeurs manquantes restantes pour Time_Orderd :", df["Time_Orderd"].isna().sum())
```
 
Cette approche évite d'introduire une heure de commande incohérente (par exemple une moyenne globale sur 24h qui ignore les pics de commande) et reste cohérente avec la variable réellement observée (`Time_Order_picked`).
 
L'indicateur `Time_Orderd_Was_Missing` est conservé comme variable candidate : il se peut que l'absence d'enregistrement de l'heure de commande soit elle-même corrélée à certaines conditions opérationnelles (panne de saisie, forte affluence, etc.).

# 36. Stratégie finale concernant les valeurs aberrantes

Les valeurs aberrantes seront traitées selon leur signification et non uniquement selon la méthode IQR.

La stratégie est :

```text
IQR outlier
      │
      ├── Valeur réellement invalide ?
      │        │
      │        └── Oui → NaN / suppression selon le contexte
      │
      └── Valeur valide mais rare ?
               │
               └── Oui → Conserver
```

Dans ce jeu de données :

* les notes `>5` seront considérées comme invalides ;
* les notes valides mais atypiques seront conservées ;
* `multiple_deliveries = 3` sera conservé ;
* les temps de livraison élevés seront conservés ;
* aucun outlier ne sera supprimé uniquement sur la base de l'IQR.

Cette approche permet d'éviter de perdre des observations importantes pour le problème métier.

## 38. Conclusion (mise à jour)
 
L'analyse complète confirme que le jeu de données est exploitable pour construire un modèle de régression destiné à prédire le temps de livraison, avec deux évolutions importantes par rapport à la version précédente du nettoyage.
 
D'abord, `Vehicle_condition` est réintégrée au jeu de variables : bien que la signification exacte de ses valeurs (0 à 3) reste non documentée, sa corrélation avec la cible est trop informative pour être ignorée. Pour éviter d'imposer une hiérarchie arbitraire, elle sera traitée comme une variable catégorielle nominale plutôt qu'ordinale.
 
Ensuite, `Time_Order_picked` est conservée afin de calculer un temps de préparation (`Preparation_Time_min`), potentiellement très prédictif du temps de livraison total. Ce choix modifie le cadrage du problème : le modèle ne prédit plus le temps de livraison au moment où la commande est passée, mais à partir du moment où elle est prise en charge par le livreur. Cette distinction doit être gardée à l'esprit lors du déploiement du modèle en conditions réelles (le use case devient une estimation en cours de trajet plutôt qu'une estimation à la commande).
 
La disponibilité de `Time_Order_picked` a également permis de résoudre le traitement des 1 731 valeurs manquantes de `Time_Orderd` : plutôt qu'une imputation arbitraire par une heure moyenne, la valeur manquante est désormais estimée à partir de l'heure de prise en charge et du temps de préparation médian observé sur les commandes complètes, une méthode plus cohérente avec la logique métier. Un indicateur binaire (`Time_Orderd_Was_Missing`) est conservé pour permettre au modèle d'exploiter une éventuelle information cachée dans l'absence de cette donnée.
 
Enfin, `Delivery_person_Age` fait l'objet d'un plafonnement (winsorisation) plutôt que d'une conservation brute : les âges inférieurs à 20 ans sont ramenés à 20, et ceux supérieurs à 40 ans sont ramenés à 40, en raison des écarts de distribution observés (peu d'observations entre 15–19 ans et entre 41–50 ans). Cette approche limite l'influence des valeurs isolées sans supprimer d'observations, dans la continuité de la stratégie déjà adoptée pour `multiple_deliveries` et `Delivery_person_Ratings`.
 
Les autres décisions du nettoyage restent inchangées : suppression des identifiants, traitement des coordonnées (0,0), conservation des outliers métier plausibles (notes valides mais atypiques, `multiple_deliveries = 3`, temps de livraison élevés), et transformation des notes > 5 en valeurs manquantes.
